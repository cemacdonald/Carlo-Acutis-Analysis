```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import arrow
from matplotlib.ticker import MaxNLocator
```

```python
# Extract events over time for each country
data = pd.read_csv('eucharistic-miracles-country.csv')
countries = ['Argentina','Austria', 'Belgium', 'Colombia', 'Croatia', 'Egypt', 'France',
             'Germany', 'Martinique', 'Reunion Islands', 'Italy', 'Mexico','Holland',
             'Peru', 'Poland', 'Portugal', 'Spain', 'Switzerland','Venezuela']

colors = cmap = plt.colormaps.get_cmap('viridis')

timeframe = {}
for country in countries:
    years = data[country].dropna().astype(int)
    timeframe[country] = [arrow.get(year, 1, 1) for year in years]

# Plot a histogram for each country
plt.figure(figsize=(18, 9))

for i, (country, events) in enumerate(timeframe.items()):
    color = colors(i / (len(countries) - 1))  # Get a unique color for each country
    plt.hist([event.year for event in events], bins=range(200,1000, 3), alpha=.75, label=country, color=color)
    plt.yticks(np.arange(0, plt.gca().get_ylim()[1]+1, 1))

# Customize the plot
plt.gca().yaxis.set_major_locator(MaxNLocator(integer=True))
plt.title('Frequency of Eurcharistic Miracles from the years 200-1000')
plt.xlabel('Year')
plt.ylabel('Frequency')
plt.legend()
plt.grid(linestyle='--', alpha=0.5)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
